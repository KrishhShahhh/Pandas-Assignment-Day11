## Applying functions to columns and creating new columns

We need to create another column in d3, which contains  a boolean value for each city to indicate whether it's a union territory or not. 
- HINT: Chandigarh, Pondicherry and Delhi are only 3 union territories here. 


```python
# write function here

def is_ut(x):
    union_territories = ['Chandigarh', 'Pondicherry', 'Delhi']
    return x in union_territories
    
#     # write code below
    
d3['is_ut'] = d3['city'].apply(is_ut)
```

```python
print(d3)

```

The temperatures mentioned in ‘Temperature’ column are mentioned in Celsius, we need another column which contains the same in Fahrenheit. 

HINT - 
- Define a function c_to_f which takes input temp in celsius and returns a value with temperature in Fahrenheit.
- To check: c_to_f(10) should return 50. 


```python
# write function here
def c_to_f(celsius):
    return (celsius * 9/5) + 32




```

```python
# check function c_to_f(10)
print(c_to_f(10))

```

```python
# apply function c_to_f to d3 to create a column 'temp_farenhiet'
d3['temp_fahrenheit'] = d3['Temperature'].apply(c_to_f)
print(d3)

```

## Groupby

```python
# Create a dataframe using dictionary of your choice

data = {
    'Department': ['HR', 'IT', 'IT', 'HR', 'Finance', 'Finance', 'IT'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace'],
    'Salary': [50000, 75000, 80000, 55000, 90000, 85000, 70000],
    'Years': [5, 3, 4, 6, 8, 7, 2]
}
df_sample = pd.DataFrame(data)
```

```python
# Use Groupby of single column with aggregate sum()
dept_salary_sum = df_sample.groupby('Department')['Salary'].sum()
print("Sum of salaries by department:\n", dept_salary_sum)
```

```python
# Use Groupby of single column with aggregate count()
dept_count = df_sample.groupby('Department')['Employee'].count()
print("\nCount of employees by department:\n", dept_count)
```

```python
# Use Groupby of single column with aggregate min() and max()
dept_salary_min_max = df_sample.groupby('Department')['Salary'].agg(['min', 'max'])
```

```python
print("\nMin and max salary by department:\n", dept_salary_min_max)
```

```python
# Use Groupby of any 2 columns with aggregate mean()

dept_years_mean = df_sample.groupby(['Department', 'Years'])['Salary'].mean()
print("\nMean salary by department and years:\n", dept_years_mean)
```

```python
# Use Groupby of any 2 columns with aggregate min() and max()
dept_years_min_max = df_sample.groupby(['Department', 'Years'])['Salary'].agg(['min', 'max'])

```

```python
print("\nMin and max salary by department and years:\n", dept_years_min_max)
```
